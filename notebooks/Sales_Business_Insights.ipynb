```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("sales_transformed.csv")
```

```python
print("Total Revenue:", df['Total'].sum())

print("Total Transactions:", df['Invoice ID'].nunique())

print("Total Quantity Sold:", df['Quantity'].sum())

print("Total Gross Income:", df['Gross income'].sum())

print("Average Transaction Value:", df['Total'].mean())
```

```python
city_sales = df.groupby('City')['Total'].sum().sort_values(ascending=False)

print(city_sales)
```

```python
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

print(product_sales)
```

```python
product_sales.plot(kind='bar', figsize=(10,5))

plt.title("Revenue by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```
