```python
import pandas as pd

df_clean = pd.read_csv("dataset/cleaned/sales_cleaned.csv")

df_clean['Date'] = pd.to_datetime(df_clean['Date'])

print("Cleaned dataset loaded successfully!")

df_clean.head()
```

```python
df_clean['Year'] = df_clean['Date'].dt.year
df_clean['Month'] = df_clean['Date'].dt.month_name()
df_clean['Day'] = df_clean['Date'].dt.day_name()
df_clean[['Date', 'Year', 'Month', 'Day']].head(10)

```

```python
df_clean['Hour'] = pd.to_datetime(
    df_clean['Time'],
    format='%H:%M',
    errors='coerce'
).dt.hour
```

```python
def categorize_time(hour):

    if pd.isna(hour):
        return "Unknown"

    elif 6 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour <= 23:
        return "Evening"

    else:
        return "Night"
```

```python
df_clean['Time Period'] = df_clean['Hour'].apply(categorize_time)
df_clean[['Time', 'Hour', 'Time Period']].head(10)
```

```python
df_clean['Profit Margin'] = (
    df_clean['Gross income'] /
    df_clean['Total']
) * 100
df_clean[['Total', 'Gross income', 'Profit Margin']].head()
```

```python
def sales_category(amount):

    if amount < 200:
        return "Low Sales"

    elif amount <= 500:
        return "Medium Sales"

    else:
        return "High Sales"
```

```python
df_clean['Sales Category'] = df_clean['Total'].apply(sales_category)
df_clean[['Total', 'Sales Category']].head(10)
```

```python
df_clean['Customer type'].value_counts()
df_clean['Customer Segment'] = df_clean['Customer type'].map({
    'Member': 'Loyal Customer',
    'Normal': 'Regular Customer'
})
df_clean[['Customer type', 'Customer Segment']].head(10)
```

```python
new_columns = [
    'Year',
    'Month',
    'Day',
    'Hour',
    'Time Period',
    'Profit Margin',
    'Sales Category',
    'Customer Segment'
]

df_clean[new_columns].head(10)
```

SET UP OF SQL

```python
import sqlite3
import pandas as pd

print("SQL environment initialized successfully!")
```

```python
df_sql = pd.read_csv("dataset/transformed/sales_transformed.csv")

print("Dataset loaded successfully!")

df_sql.head()
```

```python
connection = sqlite3.connect("sales_database.db")

print("Database connected successfully!")
```

```python
df_sql.to_sql(
    "supermarket_sales",
    connection,
    if_exists="replace",
    index=False
)

print("Sales table created successfully!")
```

```python
def run_query(query):
    result = pd.read_sql_query(query, connection)
    return result
```

```python
run_query("""
SELECT *
FROM supermarket_sales
LIMIT 10;
""")
```

```python
run_query("""
SELECT COUNT(*) AS Total_Transactions
FROM supermarket_sales;
""")
```

```python
run_query("""
SELECT ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales;
""")
```

```python
run_query("""
SELECT ROUND(AVG(Total), 2) AS Average_Sales
FROM supermarket_sales;
""")
```

```python
run_query("""
SELECT
    City,
    ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales
GROUP BY City
ORDER BY Total_Sales DESC;
""")
```

```python
run_query("""
SELECT
    "Product line",
    ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales
GROUP BY "Product line"
ORDER BY Total_Sales DESC;
""")
```

```python
run_query("""
SELECT
    Branch,
    ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales
GROUP BY Branch
ORDER BY Total_Sales DESC;
""")
```

```python
run_query("""
SELECT
    Payment,
    COUNT(*) AS Transaction_Count,
    ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales
GROUP BY Payment
ORDER BY Transaction_Count DESC;
""")
```

```python
run_query("""
SELECT
    "Customer type",
    COUNT(*) AS Total_Transactions,
    ROUND(SUM(Total), 2) AS Total_Sales,
    ROUND(AVG(Total), 2) AS Average_Sales
FROM supermarket_sales
GROUP BY "Customer type";
""")
```

```python
run_query("""
SELECT
    Month,
    ROUND(SUM(Total), 2) AS Monthly_Sales
FROM supermarket_sales
GROUP BY Month
ORDER BY MIN(Date);
""")
```

```python
run_query("""
SELECT *
FROM supermarket_sales
WHERE Total > 500;
""")
```

```python
run_query("""
SELECT
    City,
    ROUND(SUM(Total), 2) AS Total_Sales
FROM supermarket_sales
GROUP BY City
HAVING SUM(Total) > 100000;
""")
```

```python
run_query("""
SELECT

    COUNT(*) AS Total_Transactions,

    ROUND(SUM(Total), 2) AS Total_Revenue,

    ROUND(AVG(Total), 2) AS Average_Transaction_Value,

    SUM(Quantity) AS Total_Quantity_Sold,

    ROUND(SUM("Gross income"), 2) AS Total_Gross_Income

FROM supermarket_sales;
""")
```
